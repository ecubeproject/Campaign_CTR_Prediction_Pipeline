## Basic Definitions

Probability quantifies the likelihood of an event, expressed as a value between 0 (impossible) and 1 (certain).

| Term | Meaning |
|---|---|
| Sample Space (S) | The set of all possible outcomes of an experiment |
| Event (E) | A subset of the sample space — one or more outcomes of interest |
| P(E) | The likelihood that event E occurs, 0 ≤ P(E) ≤ 1 |

$$\begin{gathered}\text{Rule 1:}\quad 0 \le P(E) \le 1\\[4pt]\text{Rule 2:}\quad P(S) = 1\ \text{(probabilities of all outcomes sum to 1)}\\[4pt]\text{Rule 3 (mutually exclusive):}\quad P(E_1 \cup E_2) = P(E_1) + P(E_2)\end{gathered}$$

Two more useful flavours worth distinguishing: **empirical probability** is estimated from observed frequency (e.g. "12 of the last 100 loan applicants from Surat defaulted, so P(default) ≈ 0.12"), while **subjective probability** reflects a judgment call in the absence of repeatable data (e.g. a loan officer's gut estimate for a brand-new product line). Most ML models are built entirely on empirical probability — the frequency with which patterns occurred in training data.

## Conditional Probability & Bayes' Theorem

Conditional probability is the probability of an event given that another has already occurred — it lets you update a belief in light of new information.

$$P(A \mid B) = \frac{P(A \cap B)}{P(B)}$$

Rearranging this relationship (and its mirror, P(B|A) = P(A∩B)/P(A)) gives **Bayes' Theorem**:

$$P(A \mid B) = \frac{P(B \mid A)\cdot P(A)}{P(B)}$$

In words: *posterior* = (*likelihood* × *prior*) / *evidence*. This single formula is the entire mathematical engine behind the **Naive Bayes** classifier later in this course — it computes P(class | features) from P(features | class) and the base rate P(class), assuming features are conditionally independent.

> **💡 A worked intuition**
>
> Suppose 2% of transactions at an Ahmedabad e-commerce platform are fraudulent (the prior), and a fraud-detection rule flags 90% of actual fraud correctly but also flags 5% of legitimate transactions (false positives). If a transaction gets flagged, Bayes' Theorem is what tells you the flag is *not* 90% likely to be real fraud — because legitimate transactions vastly outnumber fraudulent ones, most flags are still false alarms. This "base rate" effect is precisely why the Naive Bayes page and the Credit Card Fraud full project both spend real time on precision, not just recall.

To calculate the actual probability that a flagged transaction is fraudulent, we use Bayes' Theorem.
## The Mathematical Formula
$$P(\text{Fraud} \mid \text{Flag}) = \frac{P(\text{Flag} \mid \text{Fraud}) \cdot P(\text{Fraud})}{P(\text{Flag})}$$
Where $P(\text{Flag})$ is the total probability of a transaction being flagged (both correctly and incorrectly):
$$P(\text{Flag}) = [P(\text{Flag} \mid \text{Fraud}) \cdot P(\text{Fraud})] + [P(\text{Flag} \mid \text{Legitimate}) \cdot P(\text{Legitimate})]$$
------------------------------
## Step-by-Step Calculation
1. Extract the Given Values

* Prior Probability of Fraud, $P(\text{Fraud})$: $2\% = 0.02$
* Prior Probability of Legitimate, $P(\text{Legitimate})$: $98\% = 0.98$
* True Positive Rate (Recall), $P(\text{Flag} \mid \text{Fraud})$: $90\% = 0.90$
* False Positive Rate, $P(\text{Flag} \mid \text{Legitimate})$: $5\% = 0.05$

2. Calculate the Total Flagged Transactions, $P(\text{Flag})$

* True Flags (Actual Fraud): $0.90 \times 0.02 = 0.018$
* False Flags (Legitimate): $0.05 \times 0.98 = 0.049$
* Total Flags: $0.018 + 0.049 = 0.067$ (or $6.7\%$ of all transactions get flagged)

3. Apply Bayes' Theorem (Calculate Precision)
$$P(\text{Fraud} \mid \text{Flag}) = \frac{0.018}{0.067} \approx 0.2687$$
------------------------------
## The Intuitive Breakdown (Using 10,000 Transactions)
Imagine the Ahmedabad e-commerce platform processes 10,000 transactions:

* 200 are fraudulent ($2\%$)
* 180 get flagged correctly ($90\%$ of 200)
   * 20 slip through undetected
* 9,800 are legitimate ($98\%$)
* 490 get flagged incorrectly ($5\%$ of 9,800)
   * 9,310 go through smoothly

If the system generates a flag, it comes from a pool of $180 + 490 = 670$ total flags.
The probability that a flagged transaction is actually fraud (Precision) is:
$$\frac{180 \text{ true fraud flags}}{670 \text{ total flags}} \approx \mathbf{26.87\%}$$
## Why Precision Matters Over Recall
Even though the system catches $90\%$ of actual fraud (Recall), roughly $73\%$ of the generated alerts are false alarms. This happens because the "base rate" of fraud is so low ($2\%$).
If a data science project focuses purely on Recall, the platform might freeze accounts for thousands of innocent users. This is why machine learning pipelines for fraud detection prioritize optimizing Precision and building models that balance this heavy class imbalance.




## Common Probability Distributions

### Normal (Gaussian) Distribution

Symmetric, bell-shaped, defined entirely by its mean (μ) and standard deviation (σ).

$$f(x) = \frac{1}{\sigma\sqrt{2\pi}}\, e^{-\frac{(x-\mu)^2}{2\sigma^2}}$$

The **Empirical Rule (68-95-99.7)**: about 68% of values fall within 1σ of the mean, 95% within 2σ, and 99.7% within 3σ. This rule is the basis for flagging statistical outliers and for the confidence-interval math on the Inferential Statistics page.

### Binomial Distribution

Models the number of successes in *n* independent trials, each with success probability *p* (e.g. "of 50 customers shown a Flipkart ad, how many click?").

$$\begin{gathered}P(X=k) = \binom{n}{k}\, p^{k}(1-p)^{n-k}\\[6pt]\text{Mean} = np \qquad \text{Variance} = np(1-p)\end{gathered}$$

### Poisson Distribution

Models the count of events in a fixed interval of time or space, given a known average rate λ (e.g. "customer support calls per hour at a Zomato helpdesk").

$$P(X=k) = \frac{e^{-\lambda}\lambda^{k}}{k!}$$

This distribution gets a full worked treatment later in **GLM & Poisson Regression**, where it becomes the basis for modelling count-valued outcomes (like Instagram Reach or store footfall) directly, instead of forcing them through ordinary linear regression.

## Permutations & Combinations

Before probabilities can be computed for many real events, you often need to count how many ways something can happen — which is exactly what the C(n,k) term inside the Binomial formula above is doing.

> **Permutations (order matters)**
>
> The number of ways to arrange *k* items out of *n*, where order matters.
> `P(n,k) = n! / (n−k)!`
> e.g. ranking the top 3 finishers out of 10 sales reps at HDFC Bank: 10×9×8 = 720 ways.

> **Combinations (order doesn't matter)**
>
> The number of ways to choose *k* items out of *n*, where order doesn't matter.
> `C(n,k) = n! / [k!(n−k)!]`
> e.g. choosing 3 stores out of 10 for a Zomato promotion pilot: C(10,3) = 120 ways.

This is precisely the C(n,k) — sometimes read "n choose k" — appearing in the Binomial pmf above: it counts the number of distinct arrangements of *k* successes among *n* trials, since "success on trials 1 and 3" and "success on trials 2 and 4" are different arrangements that each contribute separately to P(X=k).

### Geometric Distribution

Models the number of trials needed to get the *first* success, given a per-trial success probability *p* — a close cousin of the Binomial, but counting trials-until-first-success instead of successes-in-n-trials.

$$P(X=k) = (1-p)^{k-1}p \qquad \text{Mean} = \frac{1}{p}$$

e.g. if a cold-calling script for a Paytm merchant-onboarding campaign converts 1 in 5 calls (p = 0.2), the expected number of calls needed to land the first sign-up is 1/0.2 = 5 calls — useful for staffing a sales team's daily call targets.

## Joint & Conditional Distributions

Bayes' Theorem above involved single events; in practice you'll more often work with pairs (or more) of random variables at once. A **joint distribution** P(X, Y) gives the probability of two events happening together, and it's the foundation both correlation (see Inferential Statistics) and the Naive Bayes independence assumption build on.

$$P(X,Y) = P(X\mid Y)\,P(Y) = P(Y\mid X)\,P(X)$$

Two variables are **independent** exactly when P(X, Y) = P(X) · P(Y) — i.e. knowing Y tells you nothing about X. Naive Bayes gets its name from assuming this (naively) holds between every pair of features given the class label. That assumption is rarely exactly true, but it works well enough in practice to make Naive Bayes one of the most widely used classifiers for text and spam filtering.

## Beyond Normal: Heavy-Tailed Distributions

Not everything in the real world clusters neatly around a mean the way exam scores or heights do. Some phenomena follow a **power law**: a small number of extreme events account for a disproportionate share of the total — website traffic (a handful of posts get most views), city population sizes, and income distributions are classic examples.

> **💡 The "Black Swan" caution**
>
> Nassim Taleb's term for an extreme, hard-to-predict event that a normal-distribution model would treat as essentially impossible — a Flipkart Big Billion Days traffic spike, or a market crash, are exactly the kind of events that heavy-tailed real-world data can produce far more often than a Gaussian assumption would suggest. This is precisely why the Anomaly & Outlier Detection page treats "3 standard deviations from the mean" as a rule of thumb, not a guarantee — for genuinely heavy-tailed data, extreme values are simply more common than the normal distribution predicts.

## More Continuous Distributions You'll Meet Later

Normal isn't the only continuous distribution that matters. Four others recur throughout the Bayesian Approach & Gaussian Processes page and several Full Projects — each modelling a distinctly different kind of real-world uncertainty.

| Distribution | Models | Shape |
|---|---|---|
| Exponential | Time until the *next* single event, given a constant rate λ | Peaks at 0, decays smoothly — always "more likely soon than late" |
| Gamma | Time until the *k-th* event occurs (Exponential is the special case k=1) | Skewed right; shape widens and shifts as k grows |
| Beta | Uncertainty *about a probability itself* (bounded between 0 and 1) | Flexible — can look uniform, U-shaped, or peaked depending on its two parameters |
| Uniform | Total ignorance within a known range — every value equally likely | Flat rectangle between two bounds |

### Exponential — Waiting for the Next Single Event

If Zomato orders arrive in a slow zone at a constant average rate of λ = 0.2 orders/minute, the time until the *next* order follows an Exponential distribution:

````python
from scipy import stats

lam = 0.2  # orders per minute
print(1/lam)                          # 5.0  -- mean wait time (minutes)
print(stats.expon.cdf(10, scale=1/lam)) # 0.8647 -- P(next order arrives within 10 min)
````

### Gamma — Waiting for the k-th Event

If a Paytm helpdesk needs to wait for its *3rd* ticket (not just the next one) at the same underlying rate, that waiting time follows a Gamma distribution — Exponential generalised from "1st event" to "k-th event":

````python
k, theta = 3, 0.5  # shape=3 tickets, scale=0.5 hours each
print(k*theta)                       # 1.5  -- mean hours until the 3rd ticket
print(stats.gamma.cdf(2, a=k, scale=theta)) # 0.7619 -- P(3rd ticket arrives within 2 hours)
````

> **📊 Where this shows up**
>
> The Regularization page's Q&A on heteroscedastic residuals mentions modelling right-skewed positive targets (like revenue) with "a Generalised Linear Model with a Gamma distribution" — this is exactly that distribution, now with its own worked example instead of just a name-drop.

### Beta — Uncertainty About a Probability

Beta is the natural distribution for a probability you're still learning about. After observing 8 successes and 2 failures on a new Flipkart checkout flow, Beta(8, 2) describes the plausible range for the true underlying conversion rate:

````python
a, b = 8, 2   # successes, failures
print(a/(a+b))                       # 0.8000 -- mean estimate of the true conversion rate
print(stats.beta.ppf(0.025,a,b), stats.beta.ppf(0.975,a,b))
# 0.5175, 0.9719  -- 95% credible interval: true rate is plausibly anywhere in here
````

Ten observations only narrow the credible interval to roughly (0.52, 0.97) — still wide, which is exactly the honest answer after so little data. More observations tighten Beta's shape around the true rate, which is the core mechanic behind Bayesian A/B testing (see the A/B Testing Full Project).

### Uniform — Total Ignorance Within Known Bounds

If a delivery app states "your order arrives in 5 to 15 minutes" with no further information, treating every value in that range as equally likely is a Uniform distribution — already used earlier on this course for a method-of-moments estimate (θ/2 = sample mean).

````python
lo, hi = 5, 15
print((lo+hi)/2)          # 10.0 -- mean
print((hi-lo)**2/12)     # 8.3333 -- variance
````

## Chebyshev's Theorem & the Empirical Rule

How much data lies near the mean? Two rules answer this — one for *any* distribution, one specifically for the normal.

**Chebyshev's theorem** works for *any* distribution: at least `1 − 1/k²` of the data lies within `k` standard deviations of the mean (for k > 1).

````python
for k in [2, 3]:
    pct = (1 - 1 / k**2) * 100
    print(f"Chebyshev k={k}: at least {pct:.1f}% of data within {k} SD")
````

The **empirical rule (68–95–99.7)** is tighter but applies *only* to a normal distribution: about 68%, 95%, and 99.7% of data lie within 1, 2, and 3 standard deviations. We can confirm it empirically on 100,000 normal samples:

````python
import numpy as np

np.random.seed(0)
x = np.random.normal(0, 1, 100_000)
for k in [1, 2, 3]:
    print(f"within {k} SD: {np.mean(np.abs(x) < k) * 100:.1f}%")
````

The shaded bands hold ~68%, ~95%, and ~99.7% of the area under a normal curve.
